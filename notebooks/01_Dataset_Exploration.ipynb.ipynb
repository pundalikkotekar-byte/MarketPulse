```python
# Load important datasets
orders = datasets["olist_orders_dataset"]
items = datasets["olist_order_items_dataset"]
payments = datasets["olist_order_payments_dataset"]
customers = datasets["olist_customers_dataset"]
products = datasets["olist_products_dataset"]
sellers = datasets["olist_sellers_dataset"]
reviews = datasets["olist_order_reviews_dataset"]

# Convert date columns
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

# 1. Overall business performance
print("TOTAL ORDERS:", orders["order_id"].nunique())
print("DELIVERED ORDERS:", (orders["order_status"] == "delivered").sum())
print("TOTAL REVENUE: R$", round(payments["payment_value"].sum(), 2))
print("TOTAL PRODUCTS SOLD:", len(items))
print("TOTAL CUSTOMERS:", customers["customer_id"].nunique())

# 2. Monthly revenue
monthly_revenue = (
    orders[["order_id", "order_purchase_timestamp"]]
    .merge(payments.groupby("order_id")["payment_value"].sum(),
           on="order_id")
)

monthly_revenue["month"] = (
    monthly_revenue["order_purchase_timestamp"].dt.to_period("M")
)

monthly_sales = monthly_revenue.groupby("month")["payment_value"].sum()

display(monthly_sales)

# 3. Best-selling products categories
category_sales = (
    items.merge(products[["product_id", "product_category_name"]],
                on="product_id")
)

display(
    category_sales.groupby("product_category_name")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

# 4. Payment method analysis
display(payments.groupby("payment_type")["payment_value"]
        .agg(["count", "sum"])
        .sort_values("sum", ascending=False))
```

```python
# 1. Prepare order data
customer_orders = orders[
    orders["order_status"] == "delivered"
].copy()

# 2. Calculate total payment per order
order_payments = (
    payments.groupby("order_id")["payment_value"]
    .sum()
    .reset_index()
)

# 3. Connect orders with payments
customer_data = customer_orders.merge(
    order_payments,
    on="order_id",
    how="inner"
)

# 4. Add permanent customer identifiers
customer_data = customer_data.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id",
    how="left"
)

print("Customer transactions:", len(customer_data))
print("Unique customers:", customer_data["customer_unique_id"].nunique())

display(customer_data.head())
```

```python
# Reference date for recency calculation
reference_date = (
    customer_data["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

rfm = customer_data.groupby("customer_unique_id").agg(
    Recency=("order_purchase_timestamp",
             lambda x: (reference_date - x.max()).days),

    Frequency=("order_id", "nunique"),

    Monetary=("payment_value", "sum")
).reset_index()

print("RFM customer records:", len(rfm))

display(rfm.head(10))
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

display(rfm.head(10))
```

```python
def segment_customer(row):

    if row["R_Score"] >= 3 and row["F_Score"] >= 3:
        return "Loyal Customers"

    elif row["R_Score"] >= 3 and row["M_Score"] >= 3:
        return "Potential High Value"

    elif row["R_Score"] == 1 and row["F_Score"] >= 2:
        return "At Risk"

    elif row["R_Score"] == 1:
        return "Lost Customers"

    else:
        return "Regular Customers"


rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

display(
    rfm["Customer_Segment"]
    .value_counts()
    .to_frame("Customer Count")
)
```

```python
# Recency Score: more recent customers receive higher scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Frequency Score: based on actual number of purchases
rfm["F_Score"] = np.select(
    [
        rfm["Frequency"] == 1,
        rfm["Frequency"] == 2,
        rfm["Frequency"] == 3,
        rfm["Frequency"] >= 4
    ],
    [1, 2, 3, 4]
).astype(int)

# Monetary Score: higher spending receives higher scores
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Combined RFM Score
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

# Customer Segmentation
def segment_customer(row):

    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "Champions"

    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal Customers"

    elif row["R_Score"] >= 3 and row["F_Score"] == 1:
        return "Recent Customers"

    elif row["R_Score"] <= 2 and row["F_Score"] >= 2:
        return "At Risk"

    else:
        return "Hibernating Customers"


rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

# Final results
print("Total Unique Customers:", rfm["customer_unique_id"].nunique())

print("\nFrequency Distribution:")
display(rfm["Frequency"].value_counts().sort_index().to_frame("Customers"))

print("\nCustomer Segmentation:")
display(
    rfm["Customer_Segment"]
    .value_counts()
    .to_frame("Customer Count")
)

print("\nRFM Sample:")
display(rfm.head(10))
```

```python
# PRODUCT & SALES ANALYTICS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load datasets from existing dictionary
orders = datasets["olist_orders_dataset"].copy()
items = datasets["olist_order_items_dataset"].copy()
products = datasets["olist_products_dataset"].copy()
sellers = datasets["olist_sellers_dataset"].copy()
payments = datasets["olist_order_payments_dataset"].copy()
category_translation = datasets["product_category_name_translation"].copy()

# Convert dates
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

# Merge product information
product_sales = items.merge(
    products,
    on="product_id",
    how="left"
)

# Translate Portuguese category names into English
product_sales = product_sales.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

product_sales["category_english"] = (
    product_sales["product_category_name_english"]
    .fillna(product_sales["product_category_name"])
    .fillna("Unknown")
)

# Merge order dates and status
product_sales = product_sales.merge(
    orders[[
        "order_id",
        "order_purchase_timestamp",
        "order_status"
    ]],
    on="order_id",
    how="left"
)

# Delivered product sales
delivered_sales = product_sales[
    product_sales["order_status"] == "delivered"
].copy()

# Revenue by category (product price, excluding freight)
category_analysis = delivered_sales.groupby(
    "category_english"
).agg(
    Units_Sold=("order_item_id", "count"),
    Revenue=("price", "sum"),
    Average_Price=("price", "mean"),
    Total_Freight=("freight_value", "sum")
).reset_index()

category_analysis = category_analysis.sort_values(
    "Revenue",
    ascending=False
)

print("TOP 10 CATEGORIES BY REVENUE")
display(category_analysis.head(10).round(2))


# TOP 10 CATEGORIES BY UNITS SOLD
print("TOP 10 CATEGORIES BY UNITS SOLD")
display(
    category_analysis.sort_values(
        "Units_Sold",
        ascending=False
    ).head(10).round(2)
)


# SELLER PERFORMANCE
seller_analysis = delivered_sales.groupby(
    "seller_id"
).agg(
    Units_Sold=("order_item_id", "count"),
    Revenue=("price", "sum"),
    Average_Item_Price=("price", "mean")
).reset_index()

seller_analysis = seller_analysis.merge(
    sellers[["seller_id", "seller_state"]],
    on="seller_id",
    how="left"
)

seller_analysis = seller_analysis.sort_values(
    "Revenue",
    ascending=False
)

print("TOP 10 SELLERS BY REVENUE")
display(seller_analysis.head(10).round(2))


# MONTHLY SALES TREND
monthly_sales = delivered_sales.copy()

monthly_sales["Month"] = (
    monthly_sales["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

monthly_analysis = monthly_sales.groupby(
    "Month"
).agg(
    Units_Sold=("order_item_id", "count"),
    Product_Revenue=("price", "sum"),
    Freight_Revenue=("freight_value", "sum")
).reset_index()

print("MONTHLY SALES")
display(monthly_analysis.tail(15).round(2))


# AVERAGE ORDER VALUE
delivered_orders = orders[
    orders["order_status"] == "delivered"
].copy()

order_payment = payments.groupby(
    "order_id"
)["payment_value"].sum().reset_index()

delivered_order_payments = delivered_orders.merge(
    order_payment,
    on="order_id",
    how="inner"
)

aov = delivered_order_payments["payment_value"].mean()

print(f"Average Order Value: R$ {aov:,.2f}")
print(f"Delivered Orders with Payment Records: {len(delivered_order_payments):,}")
```

```python
# PRODUCT & SALES VISUALIZATIONS

# 1. Top 10 categories by revenue
top_categories = category_analysis.head(10)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_categories,
    x="Revenue",
    y="category_english"
)
plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Revenue (R$)")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# 2. Monthly product revenue
plt.figure(figsize=(14, 6))
sns.lineplot(
    data=monthly_analysis,
    x="Month",
    y="Product_Revenue",
    marker="o"
)
plt.title("Monthly Product Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Product Revenue (R$)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 3. Top 10 sellers
top_sellers = seller_analysis.head(10).copy()
top_sellers["Seller"] = top_sellers["seller_id"].str[:8]

plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_sellers,
    x="Revenue",
    y="Seller"
)
plt.title("Top 10 Sellers by Product Revenue")
plt.xlabel("Revenue (R$)")
plt.ylabel("Seller ID")
plt.tight_layout()
plt.show()
```

```python
# PAYMENT, DELIVERY & REVIEW ANALYTICS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

orders = datasets["olist_orders_dataset"].copy()
payments = datasets["olist_order_payments_dataset"].copy()
reviews = datasets["olist_order_reviews_dataset"].copy()

# Convert date columns
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")


# ==================================
# 1. PAYMENT ANALYSIS
# ==================================

payment_analysis = payments.groupby("payment_type").agg(
    Payment_Count=("order_id", "count"),
    Total_Payment=("payment_value", "sum"),
    Average_Payment=("payment_value", "mean")
).reset_index()

payment_analysis["Payment_Percentage"] = (
    payment_analysis["Payment_Count"]
    / payment_analysis["Payment_Count"].sum() * 100
).round(2)

payment_analysis = payment_analysis.sort_values(
    "Total_Payment",
    ascending=False
)

print("PAYMENT METHOD ANALYSIS")
display(payment_analysis.round(2))


# ==================================
# 2. DELIVERY ANALYSIS
# ==================================

delivered = orders[
    orders["order_status"] == "delivered"
].copy()

delivered["Actual_Delivery_Days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

delivered["Estimated_Delivery_Days"] = (
    delivered["order_estimated_delivery_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

delivered["Delivery_Difference_Days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Positive = late, negative = early
valid_delivery = delivered.dropna(
    subset=["Actual_Delivery_Days", "Delivery_Difference_Days"]
).copy()

valid_delivery["Delivery_Status"] = np.where(
    valid_delivery["Delivery_Difference_Days"] > 0,
    "Late",
    "On Time or Early"
)

delivery_summary = pd.DataFrame({
    "Metric": [
        "Delivered Orders",
        "Orders with Valid Delivery Dates",
        "Average Actual Delivery Days",
        "Average Estimated Delivery Days",
        "Average Delivery Difference (Days)",
        "Late Deliveries",
        "On-Time or Early Deliveries",
        "Late Delivery Percentage"
    ],
    "Value": [
        len(delivered),
        len(valid_delivery),
        valid_delivery["Actual_Delivery_Days"].mean(),
        valid_delivery["Estimated_Delivery_Days"].mean(),
        valid_delivery["Delivery_Difference_Days"].mean(),
        (valid_delivery["Delivery_Status"] == "Late").sum(),
        (valid_delivery["Delivery_Status"] == "On Time or Early").sum(),
        (valid_delivery["Delivery_Status"] == "Late").mean() * 100
    ]
})

print("DELIVERY PERFORMANCE")
display(delivery_summary.round(2))


# ==================================
# 3. CUSTOMER REVIEW ANALYSIS
# ==================================

review_analysis = reviews.groupby("review_score").agg(
    Review_Count=("review_id", "count")
).reset_index()

review_analysis["Review_Percentage"] = (
    review_analysis["Review_Count"]
    / review_analysis["Review_Count"].sum() * 100
).round(2)

average_review = reviews["review_score"].mean()

print("CUSTOMER REVIEW DISTRIBUTION")
display(review_analysis.round(2))

print(f"Average Customer Review Score: {average_review:.2f}/5")


# ==================================
# 4. DELIVERY VS CUSTOMER SATISFACTION
# ==================================

order_reviews = reviews.groupby("order_id").agg(
    Average_Review_Score=("review_score", "mean")
).reset_index()

delivery_reviews = valid_delivery.merge(
    order_reviews,
    on="order_id",
    how="inner"
)

delivery_satisfaction = delivery_reviews.groupby(
    "Delivery_Status"
).agg(
    Orders=("order_id", "nunique"),
    Average_Review_Score=("Average_Review_Score", "mean"),
    Average_Delivery_Days=("Actual_Delivery_Days", "mean")
).reset_index()

print("DELIVERY AND CUSTOMER SATISFACTION")
display(delivery_satisfaction.round(2))
```

```python
import pandas as pd
import os

# Load original datasets
orders = pd.read_csv("data/raw/olist_orders_dataset.csv")
payments = pd.read_csv("data/raw/olist_order_payments_dataset.csv")
customers = pd.read_csv("data/raw/olist_customers_dataset.csv")

# Keep delivered orders
customer_orders = orders[
    orders["order_status"] == "delivered"
].copy()

# Convert purchase timestamp
customer_orders["order_purchase_timestamp"] = pd.to_datetime(
    customer_orders["order_purchase_timestamp"]
)

# Aggregate payments by order
order_payments = payments.groupby("order_id")[
    "payment_value"
].sum().reset_index()

# Merge datasets
customer_data = customer_orders.merge(
    order_payments, on="order_id", how="inner"
)

customer_data = customer_data.merge(
    customers[["customer_id", "customer_unique_id"]],
    on="customer_id", how="left"
)

# Calculate RFM
reference_date = (
    customer_data["order_purchase_timestamp"].max()
    + pd.Timedelta(days=1)
)

rfm = customer_data.groupby("customer_unique_id").agg(
    Recency=(
        "order_purchase_timestamp",
        lambda x: (reference_date - x.max()).days
    ),
    Frequency=("order_id", "nunique"),
    Monetary=("payment_value", "sum")
).reset_index()

# RFM scoring
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = rfm["Frequency"].apply(
    lambda x: 1 if x == 1 else 2 if x == 2 else 3 if x == 3 else 4
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

# Customer segmentation
def segment_customer(row):
    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "Champions"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal Customers"
    elif row["R_Score"] >= 3 and row["F_Score"] == 1:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 2:
        return "At Risk"
    else:
        return "Hibernating Customers"

rfm["Customer_Segment"] = rfm.apply(
    segment_customer, axis=1
)

# Customer summary
customer_summary = rfm.groupby("Customer_Segment").agg(
    Customer_Count=("customer_unique_id", "count"),
    Average_Recency=("Recency", "mean"),
    Average_Frequency=("Frequency", "mean"),
    Average_Monetary=("Monetary", "mean"),
    Total_Monetary=("Monetary", "sum")
).round(2)

customer_summary["Customer_Percentage"] = (
    customer_summary["Customer_Count"] / len(rfm) * 100
).round(2)

# Save files
os.makedirs("data/processed", exist_ok=True)

rfm.to_csv("data/processed/customer_rfm.csv", index=False)

customer_summary.to_csv(
    "data/processed/customer_segment_summary.csv"
)

print("RFM analysis recreated successfully!")
print("Total customers:", len(rfm))
print("\nCustomer segments:")
print(customer_summary)
```
